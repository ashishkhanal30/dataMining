```python
# Import visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime

# Set style for better-looking plots
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

```python
# Convert crash date to datetime and extract time components
df['CRASH DATE'] = pd.to_datetime(df['CRASH DATE'])
df['Year'] = df['CRASH DATE'].dt.year
df['Month'] = df['CRASH DATE'].dt.month
df['DayOfWeek'] = df['CRASH DATE'].dt.day_name()
df['Hour'] = df['CRASH DATE'].dt.hour

# Create a comprehensive overview of crash patterns
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Baton Rouge Traffic Crash Patterns Overview', fontsize=16, fontweight='bold')

# 1. Crashes by Month
monthly_crashes = df.groupby('Month').size()
axes[0, 0].bar(monthly_crashes.index, monthly_crashes.values, color='skyblue', alpha=0.7)
axes[0, 0].set_title('Crashes by Month')
axes[0, 0].set_xlabel('Month')
axes[0, 0].set_ylabel('Number of Crashes')
axes[0, 0].set_xticks(range(1, 13))
axes[0, 0].set_xticklabels(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
                           'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])

# 2. Crashes by Day of Week
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_crashes = df['DayOfWeek'].value_counts().reindex(day_order)
axes[0, 1].bar(range(len(daily_crashes)), daily_crashes.values, color='lightcoral', alpha=0.7)
axes[0, 1].set_title('Crashes by Day of Week')
axes[0, 1].set_xlabel('Day of Week')
axes[0, 1].set_ylabel('Number of Crashes')
axes[0, 1].set_xticks(range(len(daily_crashes)))
axes[0, 1].set_xticklabels([day[:3] for day in daily_crashes.index], rotation=45)

# 3. Crashes by Hour of Day
hourly_crashes = df.groupby('Hour').size()
axes[1, 0].plot(hourly_crashes.index, hourly_crashes.values, marker='o', linewidth=2, markersize=4)
axes[1, 0].set_title('Crashes by Hour of Day')
axes[1, 0].set_xlabel('Hour')
axes[1, 0].set_ylabel('Number of Crashes')
axes[1, 0].grid(True, alpha=0.3)

# 4. Crash Severity Distribution
severity_data = df[['INJURY', 'FATALITY']].sum()
axes[1, 1].pie(severity_data.values, labels=['Injuries', 'Fatalities'], autopct='%1.1f%%', 
               colors=['lightgreen', 'salmon'])
axes[1, 1].set_title('Crash Severity Distribution')

plt.tight_layout()
plt.show()

print(f"Total crashes analyzed: {len(df):,}")
print(f"Date range: {df['CRASH DATE'].min().strftime('%Y-%m-%d')} to {df['CRASH DATE'].max().strftime('%Y-%m-%d')}")
print(f"Total injuries: {df['INJURY'].sum():,}")
print(f"Total fatalities: {df['FATALITY'].sum():,}")
```

```python
# Detailed analysis of crash severity and contributing factors
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Crash Severity and Contributing Factors Analysis', fontsize=16, fontweight='bold')

# 1. Weather Conditions vs Crash Severity
weather_injury = df.groupby('WEATHER CONDITION')['INJURY'].sum().sort_values(ascending=False)
weather_fatality = df.groupby('WEATHER CONDITION')['FATALITY'].sum().sort_values(ascending=False)

# Top 8 weather conditions for better readability
top_weather = weather_injury.head(8).index
weather_injury_top = weather_injury[top_weather]
weather_fatality_top = weather_fatality[top_weather]

x = np.arange(len(top_weather))
width = 0.35

axes[0, 0].bar(x - width/2, weather_injury_top.values, width, label='Injuries', alpha=0.8, color='orange')
axes[0, 0].bar(x + width/2, weather_fatality_top.values, width, label='Fatalities', alpha=0.8, color='red')
axes[0, 0].set_title('Weather Conditions vs Crash Severity')
axes[0, 0].set_xlabel('Weather Condition')
axes[0, 0].set_ylabel('Count')
axes[0, 0].set_xticks(x)
axes[0, 0].set_xticklabels([w[:15] + '...' if len(w) > 15 else w for w in top_weather], rotation=45, ha='right')
axes[0, 0].legend()

# 2. Lighting Conditions Analysis
lighting_counts = df['LIGHTING CONDITION'].value_counts()
axes[0, 1].pie(lighting_counts.values, labels=lighting_counts.index, autopct='%1.1f%%', startangle=90)
axes[0, 1].set_title('Distribution of Lighting Conditions')

# 3. Road Surface Conditions
road_surface = df['ROAD SURFACE'].value_counts()
axes[1, 0].barh(range(len(road_surface)), road_surface.values, color='lightblue', alpha=0.7)
axes[1, 0].set_title('Road Surface Conditions')
axes[1, 0].set_xlabel('Number of Crashes')
axes[1, 0].set_yticks(range(len(road_surface)))
axes[1, 0].set_yticklabels(road_surface.index)

# 4. Hit and Run vs Regular Crashes Severity
hit_run_severity = df.groupby('HIT AND RUN')[['INJURY', 'FATALITY']].sum()
hit_run_labels = ['Regular Crash', 'Hit and Run']
hit_run_injuries = hit_run_severity['INJURY'].values
hit_run_fatalities = hit_run_severity['FATALITY'].values

x = np.arange(len(hit_run_labels))
width = 0.35

axes[1, 1].bar(x - width/2, hit_run_injuries, width, label='Injuries', alpha=0.8, color='lightgreen')
axes[1, 1].bar(x + width/2, hit_run_fatalities, width, label='Fatalities', alpha=0.8, color='darkred')
axes[1, 1].set_title('Hit and Run vs Regular Crashes Severity')
axes[1, 1].set_xlabel('Crash Type')
axes[1, 1].set_ylabel('Count')
axes[1, 1].set_xticks(x)
axes[1, 1].set_xticklabels(hit_run_labels)
axes[1, 1].legend()

plt.tight_layout()
plt.show()

# Print some key statistics
print("Key Statistics:")
print(f"Hit and Run crashes: {df['HIT AND RUN'].sum():,} ({df['HIT AND RUN'].mean()*100:.1f}% of all crashes)")
print(f"Pedestrian involved crashes: {df['PEDESTRIAN INVOLVED'].sum():,} ({df['PEDESTRIAN INVOLVED'].mean()*100:.1f}% of all crashes)")
print(f"Train involved crashes: {df['TRAIN INVOLVED'].sum():,} ({df['TRAIN INVOLVED'].mean()*100:.1f}% of all crashes)")
print(f"Intersection crashes: {df['INTERSECTION'].sum():,} ({df['INTERSECTION'].mean()*100:.1f}% of all crashes)")
```

```python
# Geographic visualization of crash locations
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Geographic Distribution of Traffic Crashes in Baton Rouge', fontsize=16, fontweight='bold')

# 1. All crashes scatter plot
axes[0, 0].scatter(df['LONGITUDE'], df['LATITUDE'], alpha=0.3, s=1, color='blue')
axes[0, 0].set_title('All Traffic Crashes - Geographic Distribution')
axes[0, 0].set_xlabel('Longitude')
axes[0, 0].set_ylabel('Latitude')
axes[0, 0].grid(True, alpha=0.3)

# 2. Crashes with injuries
injury_crashes = df[df['INJURY'] > 0]
axes[0, 1].scatter(injury_crashes['LONGITUDE'], injury_crashes['LATITUDE'], 
                   alpha=0.5, s=2, color='orange', label=f'Injury Crashes ({len(injury_crashes):,})')
axes[0, 1].set_title('Crashes with Injuries')
axes[0, 1].set_xlabel('Longitude')
axes[0, 1].set_ylabel('Latitude')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].legend()

# 3. Fatal crashes
fatal_crashes = df[df['FATALITY'] > 0]
axes[1, 0].scatter(fatal_crashes['LONGITUDE'], fatal_crashes['LATITUDE'], 
                   alpha=0.8, s=10, color='red', label=f'Fatal Crashes ({len(fatal_crashes):,})')
axes[1, 0].set_title('Fatal Crashes')
axes[1, 0].set_xlabel('Longitude')
axes[1, 0].set_ylabel('Latitude')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].legend()

# 4. Hit and Run crashes
hit_run_crashes = df[df['HIT AND RUN'] == 1]
axes[1, 1].scatter(hit_run_crashes['LONGITUDE'], hit_run_crashes['LATITUDE'], 
                   alpha=0.5, s=2, color='purple', label=f'Hit and Run ({len(hit_run_crashes):,})')
axes[1, 1].set_title('Hit and Run Crashes')
axes[1, 1].set_xlabel('Longitude')
axes[1, 1].set_ylabel('Latitude')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].legend()

plt.tight_layout()
plt.show()

# Geographic statistics
print("Geographic Analysis:")
print(f"Latitude range: {df['LATITUDE'].min():.4f} to {df['LATITUDE'].max():.4f}")
print(f"Longitude range: {df['LONGITUDE'].min():.4f} to {df['LONGITUDE'].max():.4f}")
print(f"Crashes with valid coordinates: {df[df['LATITUDE'].notna() & df['LONGITUDE'].notna()].shape[0]:,}")
print(f"Percentage of crashes with injuries: {len(injury_crashes)/len(df)*100:.1f}%")
print(f"Percentage of fatal crashes: {len(fatal_crashes)/len(df)*100:.1f}%")
print(f"Percentage of hit and run crashes: {len(hit_run_crashes)/len(df)*100:.1f}%")
```

```python
# Detailed analysis of crash factors and primary/secondary causes
fig, axes = plt.subplots(2, 2, figsize=(18, 14))
fig.suptitle('Crash Factors and Contributing Causes Analysis', fontsize=16, fontweight='bold')

# 1. Primary Factors causing crashes
primary_factors = df['PRIMARY FACTOR'].value_counts().head(10)
axes[0, 0].barh(range(len(primary_factors)), primary_factors.values, color='lightcoral', alpha=0.7)
axes[0, 0].set_title('Top 10 Primary Factors in Crashes')
axes[0, 0].set_xlabel('Number of Crashes')
axes[0, 0].set_yticks(range(len(primary_factors)))
axes[0, 0].set_yticklabels([factor[:30] + '...' if len(factor) > 30 else factor for factor in primary_factors.index])

# 2. Secondary Factors
secondary_factors = df['SECONDARY FACTOR'].value_counts().head(10)
axes[0, 1].barh(range(len(secondary_factors)), secondary_factors.values, color='lightblue', alpha=0.7)
axes[0, 1].set_title('Top 10 Secondary Factors in Crashes')
axes[0, 1].set_xlabel('Number of Crashes')
axes[0, 1].set_yticks(range(len(secondary_factors)))
axes[0, 1].set_yticklabels([factor[:30] + '...' if len(factor) > 30 else factor for factor in secondary_factors.index])

# 3. Manner of Collision
collision_types = df['MANNER OF COLLISION'].value_counts()
axes[1, 0].pie(collision_types.values, labels=collision_types.index, autopct='%1.1f%%', startangle=90)
axes[1, 0].set_title('Distribution of Collision Types')

# 4. Vehicle Alignment Analysis
vehicle_alignment = df['VEHICLE ALIGNMENT'].value_counts()
axes[1, 1].bar(range(len(vehicle_alignment)), vehicle_alignment.values, color='lightgreen', alpha=0.7)
axes[1, 1].set_title('Vehicle Alignment Distribution')
axes[1, 1].set_xlabel('Vehicle Alignment')
axes[1, 1].set_ylabel('Number of Crashes')
axes[1, 1].set_xticks(range(len(vehicle_alignment)))
axes[1, 1].set_xticklabels([align[:15] + '...' if len(align) > 15 else align for align in vehicle_alignment.index], rotation=45, ha='right')

plt.tight_layout()
plt.show()

# Print detailed statistics
print("Crash Factors Analysis:")
print(f"Most common primary factor: {primary_factors.index[0]} ({primary_factors.iloc[0]:,} crashes)")
print(f"Most common secondary factor: {secondary_factors.index[0]} ({secondary_factors.iloc[0]:,} crashes)")
print(f"Most common collision type: {collision_types.index[0]} ({collision_types.iloc[0]:,} crashes)")
print(f"Most common vehicle alignment: {vehicle_alignment.index[0]} ({vehicle_alignment.iloc[0]:,} crashes)")
```

```python
# Vehicle-related analysis and district/zone patterns
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Vehicle and Location Analysis', fontsize=16, fontweight='bold')

# 1. Total Vehicles Distribution
vehicle_counts = df['TOTAL VEHICLES'].value_counts().sort_index()
axes[0, 0].bar(vehicle_counts.index, vehicle_counts.values, color='skyblue', alpha=0.7)
axes[0, 0].set_title('Distribution of Total Vehicles per Crash')
axes[0, 0].set_xlabel('Number of Vehicles')
axes[0, 0].set_ylabel('Number of Crashes')
axes[0, 0].grid(True, alpha=0.3)

# 2. District Analysis
district_crashes = df['DISTRICT'].value_counts().sort_index()
axes[0, 1].bar(district_crashes.index, district_crashes.values, color='lightcoral', alpha=0.7)
axes[0, 1].set_title('Crashes by District')
axes[0, 1].set_xlabel('District')
axes[0, 1].set_ylabel('Number of Crashes')
axes[0, 1].grid(True, alpha=0.3)

# 3. Zone Analysis (Top 15 zones)
zone_crashes = df['ZONE'].value_counts().head(15)
axes[1, 0].barh(range(len(zone_crashes)), zone_crashes.values, color='lightgreen', alpha=0.7)
axes[1, 0].set_title('Top 15 Zones by Crash Count')
axes[1, 0].set_xlabel('Number of Crashes')
axes[1, 0].set_yticks(range(len(zone_crashes)))
axes[1, 0].set_yticklabels([zone[:20] + '...' if len(zone) > 20 else zone for zone in zone_crashes.index])

# 4. Relation to Roadway
roadway_relation = df['RELATION TO ROADWAY'].value_counts()
axes[1, 1].pie(roadway_relation.values, labels=roadway_relation.index, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title('Relation to Roadway Distribution')

plt.tight_layout()
plt.show()

# Vehicle and location statistics
print("Vehicle and Location Analysis:")
print(f"Average vehicles per crash: {df['TOTAL VEHICLES'].mean():.2f}")
print(f"Maximum vehicles in a single crash: {df['TOTAL VEHICLES'].max()}")
print(f"Most crash-prone district: District {district_crashes.index[0]} ({district_crashes.iloc[0]:,} crashes)")
print(f"Most crash-prone zone: {zone_crashes.index[0]} ({zone_crashes.iloc[0]:,} crashes)")
print(f"Total number of districts: {df['DISTRICT'].nunique()}")
print(f"Total number of zones: {df['ZONE'].nunique()}")
```

```python
# Correlation heatmap and summary statistics
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Correlation Analysis and Summary Statistics', fontsize=16, fontweight='bold')

# 1. Correlation heatmap for numerical variables
numerical_cols = ['DISTRICT', 'INJURY', 'FATALITY', 'HIT AND RUN', 'PEDESTRIAN INVOLVED', 
                  'TRAIN INVOLVED', 'TOTAL VEHICLES', 'LATITUDE', 'LONGITUDE']
correlation_matrix = df[numerical_cols].corr()

im = axes[0].imshow(correlation_matrix, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)
axes[0].set_title('Correlation Matrix of Numerical Variables')
axes[0].set_xticks(range(len(numerical_cols)))
axes[0].set_yticks(range(len(numerical_cols)))
axes[0].set_xticklabels([col[:10] + '...' if len(col) > 10 else col for col in numerical_cols], rotation=45, ha='right')
axes[0].set_yticklabels([col[:10] + '...' if len(col) > 10 else col for col in numerical_cols])

# Add correlation values as text
for i in range(len(numerical_cols)):
    for j in range(len(numerical_cols)):
        text = axes[0].text(j, i, f'{correlation_matrix.iloc[i, j]:.2f}',
                           ha="center", va="center", color="black", fontsize=8)

# Add colorbar
plt.colorbar(im, ax=axes[0])

# 2. Summary statistics table
summary_stats = df[['INJURY', 'FATALITY', 'HIT AND RUN', 'PEDESTRIAN INVOLVED', 
                   'TRAIN INVOLVED', 'TOTAL VEHICLES']].describe().round(2)

# Create a text table
axes[1].axis('tight')
axes[1].axis('off')
table_data = []
for col in summary_stats.columns:
    table_data.append([col] + [str(summary_stats.loc[stat, col]) for stat in summary_stats.index])

table = axes[1].table(cellText=table_data[1:], colLabels=['Variable'] + list(summary_stats.index),
                     cellLoc='center', loc='center')
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.2, 1.5)
axes[1].set_title('Summary Statistics')

plt.tight_layout()
plt.show()

# Final comprehensive summary
print("\n" + "="*60)
print("COMPREHENSIVE TRAFFIC CRASH ANALYSIS SUMMARY")
print("="*60)
print(f"Dataset Overview:")
print(f"  • Total crashes analyzed: {len(df):,}")
print(f"  • Date range: {df['CRASH DATE'].min().strftime('%Y-%m-%d')} to {df['CRASH DATE'].max().strftime('%Y-%m-%d')}")
print(f"  • Geographic coverage: {df['DISTRICT'].nunique()} districts, {df['ZONE'].nunique()} zones")

print(f"\nCrash Severity:")
print(f"  • Crashes with injuries: {df['INJURY'].sum():,} ({df['INJURY'].mean()*100:.1f}% of all crashes)")
print(f"  • Fatal crashes: {df['FATALITY'].sum():,} ({df['FATALITY'].mean()*100:.1f}% of all crashes)")
print(f"  • Hit and run crashes: {df['HIT AND RUN'].sum():,} ({df['HIT AND RUN'].mean()*100:.1f}% of all crashes)")

print(f"\nSpecial Circumstances:")
print(f"  • Pedestrian involved: {df['PEDESTRIAN INVOLVED'].sum():,} ({df['PEDESTRIAN INVOLVED'].mean()*100:.1f}%)")
print(f"  • Train involved: {df['TRAIN INVOLVED'].sum():,} ({df['TRAIN INVOLVED'].mean()*100:.1f}%)")
print(f"  • Intersection crashes: {df['INTERSECTION'].sum():,} ({df['INTERSECTION'].mean()*100:.1f}%)")

print(f"\nVehicle Statistics:")
print(f"  • Average vehicles per crash: {df['TOTAL VEHICLES'].mean():.2f}")
print(f"  • Most common vehicle count: {df['TOTAL VEHICLES'].mode().iloc[0]} vehicles")

print(f"\nTemporal Patterns:")
print(f"  • Peak crash month: {df['Month'].mode().iloc[0]} ({df.groupby('Month').size().max():,} crashes)")
print(f"  • Peak crash day: {df['DayOfWeek'].mode().iloc[0]} ({df['DayOfWeek'].value_counts().max():,} crashes)")
print(f"  • Peak crash hour: {df['Hour'].mode().iloc[0]}:00 ({df.groupby('Hour').size().max():,} crashes)")
```
