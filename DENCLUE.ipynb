```python
x = [91, 106, 120, 100, 110, 125, 141, 152, 130, 140]
y = [186, 198, 220, 210, 190, 225, 258, 237, 230, 240]
sigma = 22.1
import numpy as np

# Square influence function
a = []
for i in range(len(x)):
    d = []
    for j in range(len(y)):
        if i == j:
            d.append(0.0)
        else:
            d.append(float(max(abs(x[j] - x[i]), abs(y[j] - y[i]))))
    a.append(d)
a = np.array(a)
print(a)
# Density function
density = []
for i in range(len(a)):
    temp = 0
    for j in range(len(a[i])):
        if a[i][j] > sigma:
            temp += 0
        else:
            temp += 1
    density.append(temp)
density = np.array(density)
            
# print(density)
```

```python
n = len(a)
n
```

```python
def find_attractor(i):
    current = i
    while True:
        # find neighbors within sigma
        neighbors = np.where(a[current] <= sigma)[0]

        # get densities of neighbors

        print(neighbors, '-'*10)
        neighbor_dens = density[neighbors]

        # highest density among neighbors
        max_dens = neighbor_dens.max()

        # if current point already has max density → attractor reached
        if density[current] == max_dens:
            return current
        
        # otherwise move to neighbor with highest density
        # if several, just pick the first
        next_point = neighbors[np.argmax(neighbor_dens)]
        current = next_point


# compute attractor for each point
attractors = np.array([find_attractor(i) for i in range(n)])
print("Attractor index for each point:", attractors.tolist())

# ---------------------------
# Step 4 — Cluster formation
# Points with same attractor belong to same cluster.
# ---------------------------

# unique attractor points
unique_attr = np.unique(attractors)

# assign cluster labels 0,1,2...
cluster_map = {attr: idx for idx, attr in enumerate(unique_attr)}

# label of each point = cluster of its attractor
labels = np.array([cluster_map[a] for a in attractors])
print("Cluster labels:", labels.tolist())
```
