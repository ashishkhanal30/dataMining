#### Imputing column intersecting street

Getting the road name using the property latitude and longitude as there are 0 empty values for this. To retrieve the road name, a package geopy is used, and the values for intersecting street are populated for all 175 empty records. 

```python
# first 5 rows where 'INTERSECTING STREET' is null 
df[df['INTERSECTING STREET'].isnull()].head()
```

```python
geolocator = Nominatim(user_agent="study_crash_pattern")
```

```python
def get_intersecting_street(row):
    try:
        location = geolocator.reverse(f"{row['LATITUDE']}, {row['LONGITUDE']}", exactly_one=True)
        address = location.raw.get('address', {})
        return address.get('road', 'Not Found') 
    except Exception as e:
        return 'Error'
    finally:
        time.sleep(1)
```

```python
# Create a filter for missing value
missing_mask = df['INTERSECTING STREET'].isnull()

# Apply .loc accessor to select and update the respective value after applying the function get_intersecting_street
df.loc[missing_mask, 'INTERSECTING STREET'] = df[missing_mask].apply(get_intersecting_street, axis=1)
```

#### Updating weather condition values for simplicity

```python
df['WEATHER CONDITION'].unique()
```

```python
# Convert all entries to lowercase and remove leading/trailing whitespace
df['WEATHER CONDITION'] = df['WEATHER CONDITION'].str.lower().str.strip()
```

```python
# Define a mapping from the old, complex values to the new, clean ones
weather_mapping = {
    'fog, smog, smoke': 'fog',
    'freezing rain or freezing drizzle': 'freezing rain',
    'blowing snow': 'snow'
    # 'severe crosswinds' could be grouped into 'other' if you want fewer categories
}

# Apply the replacement to the column
df['WEATHER CONDITION'].replace(weather_mapping, inplace=True)
```

#### Encode all categorical values using one-hot encoding

```python
# Get high cardinality columns
print(df.select_dtypes(include=['object']).nunique())
```

```python
# Drop columns that are identifier, have high cardinality values 

columns_drop = [
    'INCIDENT NUMBER', 
    'STREET ADDRESS', 
    'INTERSECTING STREET', 
    'GEOMETRY', 
    'STATE'
]

df_cleaned = df.drop(columns=columns_drop)
```

```python
# Calculate the cumulative percentage contribution of each zone
zone_distribution = df_cleaned['ZONE'].value_counts(normalize=True).cumsum()

# Use .to_string() to ensure the full list prints out
# print(zone_distribution.to_string())
```

```python
N = 40
top_n_zones = df_cleaned['ZONE'].value_counts().nlargest(N).index
df_cleaned['ZONE'] = df_cleaned['ZONE'].where(df_cleaned['ZONE'].isin(top_n_zones), 'Other')
```

```python
df_cleaned['CRASH DATE'] = df_cleaned['CRASH DATE'].dt.date
```

```python
original_columns = df_cleaned.columns
clean_cols = [col.lower().replace(' ', '_') for col in original_columns]
df_cleaned.columns = clean_cols
```

```python
# Grouping collision types
df_cleaned['manner_of_collision'] = df_cleaned['manner_of_collision'].str.lower()

collision_mapping = {
    # Rear-End
    'front to rear - rear end': 'Rear-End',
    
    # Sideswipe
    'sideswipe - same direction': 'Sideswipe',
    'sideswipe - opposite direction': 'Sideswipe',
    'sideswipe - left overtake': 'Sideswipe',
    'sideswipe - right overtake': 'Sideswipe',
    
    # Angle
    'angle - perpendicular/other angle': 'Angle',
    'angle - left opposite direction': 'Angle',
    'angle - left into flow': 'Angle',
    'angle - right into flow': 'Angle',
    'angle - left across flow': 'Angle',
    'angle - right across flow': 'Angle',
    'angle - left overtake': 'Angle',
    'angle - right overtake': 'Angle',

    # Backing
    'backing - rear to front': 'Backing',
    'backing - rear to side': 'Backing',
    'backing - rear to rear': 'Backing',
    
    # Head-On
    'front to front - head on': 'Head-On',
    'front to front - left against flow': 'Head-On',
    'front to front - right against flow': 'Head-On',
    
    # Single Vehicle
    'not a collision between two motor vehicles in transport': 'Single Vehicle / Non-Collision',
    
    # Other / Unknown
    'other': 'Other',
    'unknown': 'Unknown'
}

# Apply the replacement to the column
df_cleaned['manner_of_collision'] = df_cleaned['manner_of_collision'].replace(collision_mapping)
```
